```python
import os

import pandas as pd
import requests
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
api_key = os.environ["COINGECKO_KEY"]

url = "https://api.coingecko.com/api/v3/simple/price?ids={id}"
coins_ids = ["bitcoin", "ethereum", "tether", "solana", "cardano"]
params = {
    "vs_currencies": "usd",
    "x_cg_demo_api_key": api_key
}
```

## Puxando os dados da API

```python
raw_data = []
for coin_id in coins_ids:
    url_loop = url.format(id=coin_id)
    response = requests.get(url_loop, params=params)
    raw_data.append((coin_id, response.json()))
```

```python
print(raw_data)
```

# Transformando em DataFrame

```python
rows = []
for coin_id, data in raw_data:
    coin_data = data.get(coin_id, {})  # pega o dict interno, ou {} se vier vazio
    rows.append({
        "coin_id": coin_id,
        "price_usd": coin_data.get("usd")
    })

df = pd.DataFrame(rows)
```
